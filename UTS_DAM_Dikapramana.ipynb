```python
# STEP 11: AKUISISI DATA API OPEN-METEO

# URL API cuaca historis
url_api = "https://archive-api.open-meteo.com/v1/archive"

# Email kampus untuk identitas User-Agent
# Ganti dengan email kampus yang sebenarnya sebelum menjalankan kode
EMAIL_KAMPUS = "dikapramana@instiki.ac.id"

headers_api = {
    "User-Agent": f"UTS-DAM-Informatika/1.0 (mailto:{"dikapramana@instiki.ac.id"})"
}

# Parameter pengambilan data cuaca Kabupaten Malang
params_api = {
    "latitude": -8.10,
    "longitude": 112.70,
    "start_date": "2026-10-15",
    "end_date": "2026-11-13",
    "daily": [
        "temperature_2m_mean",
        "temperature_2m_max",
        "temperature_2m_min",
        "precipitation_sum",
        "wind_speed_10m_max",
        "relative_humidity_2m_mean"
    ],
    "timezone": "Asia/Jakarta"
}

# Menyiapkan percobaan request dengan retry
maksimal_percobaan = 3
data_api = None

for percobaan in range(1, maksimal_percobaan + 1):

    print(f"Percobaan ke-{percobaan} mengakses Open-Meteo...")

    try:
        # Mengirim request dengan batas waktu 30 detik
        response_api = requests.get(
            url_api,
            params=params_api,
            headers=headers_api,
            timeout=30
        )

        # Memeriksa status HTTP
        print("Status HTTP:", response_api.status_code)
        response_api.raise_for_status()

        # Membaca respons JSON
        data_api = response_api.json()

        print("Pengambilan data berhasil!")
        break

    except requests.exceptions.RequestException as e:
        print("Request gagal:", e)

        # Jeda sebelum mencoba kembali
        if percobaan < maksimal_percobaan:
            print("Menunggu 3 detik sebelum mencoba kembali...")
            time.sleep(3)

    except ValueError:
        print("Respons API bukan JSON yang valid.")
        break

# Memproses data jika API berhasil
if data_api is not None:

    # Mengubah data harian JSON menjadi DataFrame
    df_api_cuaca = pd.DataFrame(data_api["daily"])

    # Menyimpan respons mentah tanpa perubahan
    file_raw_api = RAW_DIR / "open_meteo_malang_raw.json"

    with open(file_raw_api, "w", encoding="utf-8") as f:
        json.dump(data_api, f, indent=4, ensure_ascii=False)

    print("\n=== HASIL AKUISISI API ===")
    print("Jumlah baris:", len(df_api_cuaca))
    print("Jumlah kolom:", len(df_api_cuaca.columns))
    print("Kolom:", df_api_cuaca.columns.tolist())
    print("File JSON mentah:", file_raw_api)

    display(df_api_cuaca.head())

else:
    print("\nAPI belum berhasil diakses setelah 3 percobaan.")
```

```python
# STEP 14: UJI AKSES WEBSITE DAN ROBOTS.TXT

# URL halaman daftar hotel Malang
url_scraping = "https://www.east-java.com/tourism/malang/beach/hotels.html"

# URL robots.txt website
url_robots = "https://www.east-java.com/robots.txt"

# Menggunakan User-Agent yang sudah disiapkan sebelumnya
headers_scraping = headers_api

try:
    # Memeriksa robots.txt terlebih dahulu
    response_robots = requests.get(
        url_robots,
        headers=headers_scraping,
        timeout=30
    )

    print("=== PEMERIKSAAN ROBOTS.TXT ===")
    print("Status HTTP:", response_robots.status_code)
    print("URL:", url_robots)
    print(response_robots.text[:2000])

    # Memberikan jeda sebelum mengakses halaman
    time.sleep(3)

    # Memeriksa akses halaman hotel
    response_scraping = requests.get(
        url_scraping,
        headers=headers_scraping,
        timeout=30
    )

    print("\n=== PEMERIKSAAN HALAMAN HOTEL ===")
    print("Status HTTP:", response_scraping.status_code)
    print("URL:", response_scraping.url)
    print("Content-Type:", response_scraping.headers.get("Content-Type"))
    print("Ukuran halaman:", len(response_scraping.content), "byte")

except requests.exceptions.RequestException as e:
    print("Terjadi kendala koneksi:", e)
```

```python
# STEP 15: PEMERIKSAAN STRUKTUR HTML

# Menggunakan HTML yang sudah berhasil diperoleh pada Cell 14
soup_scraping = BeautifulSoup(
    response_scraping.text,
    "html.parser"
)

# Memeriksa judul halaman
print("=== JUDUL HALAMAN ===")
print(soup_scraping.title.get_text(strip=True) if soup_scraping.title else "Judul tidak ditemukan")

# Memeriksa heading halaman
print("\n=== HEADING HALAMAN ===")

for heading in soup_scraping.find_all(["h1", "h2", "h3"]):
    print(heading.get_text(" ", strip=True))

# Mengambil teks halaman untuk pemeriksaan awal
teks_halaman = soup_scraping.get_text(" ", strip=True)

print("\n=== PREVIEW ISI HALAMAN ===")
print(teks_halaman[:3000])

# Menghitung jumlah elemen HTML
print("\n=== STRUKTUR HTML ===")
print("Jumlah tabel:", len(soup_scraping.find_all("table")))
print("Jumlah link:", len(soup_scraping.find_all("a")))
print("Jumlah gambar:", len(soup_scraping.find_all("img")))
```

```python
# STEP 16: EKSTRAKSI DATA HOTEL DARI HTML

# Mengambil seluruh teks dari halaman HTML
teks_hotel = soup_scraping.get_text("\n", strip=True)

# Membagi teks menjadi baris-baris
baris_hotel = teks_hotel.split("\n")

# Mengambil baris yang berisi informasi hotel
data_hotel_scraping = []

for baris in baris_hotel:
    baris = baris.strip()

    # Mengabaikan baris kosong dan menu website
    if baris and (
        "Hotel" in baris or
        "Inn" in baris or
        "Wisata" in baris or
        "Homestay" in baris
    ):
        data_hotel_scraping.append(baris)

# Menghapus data duplikat tanpa mengubah urutan
data_hotel_scraping = list(dict.fromkeys(data_hotel_scraping))

# Menampilkan hasil ekstraksi
print("=== HASIL EKSTRAKSI HOTEL ===")
print("Jumlah data ditemukan:", len(data_hotel_scraping))

for i, nama in enumerate(data_hotel_scraping, start=1):
    print(f"{i}. {nama}")
```

```python
# STEP 17: MEMERIKSA STRUKTUR HTML HOTEL

# Mencari teks hotel pertama yang ditemukan
target_hotel = soup_scraping.find(
    string=lambda teks: teks and "Pelangi" in teks
)

print("=== PEMERIKSAAN STRUKTUR HOTEL ===")

if target_hotel:
    elemen = target_hotel.parent

    # Menampilkan struktur elemen HTML di sekitar nama hotel
    for tingkat in range(5):
        if elemen:
            print(f"\nTingkat {tingkat + 1}:")
            print("Tag:", elemen.name)
            print("Class:", elemen.get("class"))
            print("ID:", elemen.get("id"))
            print("Isi:", elemen.get_text(" ", strip=True)[:500])

            elemen = elemen.parent
else:
    print("Nama hotel tidak ditemukan di struktur HTML.")
```

```python

# STEP 33: SCRAPING HOTEL KABUPATEN MALANG DARI TIMES MALANG
# Mengambil daftar hotel dari halaman 1, 2, dan 3

import requests
import pandas as pd
import time
from bs4 import BeautifulSoup
from datetime import datetime

url_halaman = [
    "https://timesmalang.com/penginapan/hotel",
    "https://timesmalang.com/penginapan/hotel/halaman-2",
    "https://timesmalang.com/penginapan/hotel/halaman-3"
]

data_hotel_baru = []

for url in url_halaman:
    time.sleep(2)

    response = requests.get(
        url,
        headers=headers,
        timeout=30
    )

    print("Memeriksa:", url)
    print("Status HTTP:", response.status_code)

    if response.status_code == 200:
        soup = BeautifulSoup(response.text, "html.parser")

        for link in soup.find_all("a", href=True):
            nama = link.get_text(" ", strip=True)
            alamat = link["href"]

            if "/tempat/" in alamat and nama:
                teks = nama.lower()

                if "kabupaten malang" in teks:
                    data_hotel_baru.append({
                        "Nama_Hotel": nama,
                        "URL_Detail": alamat
                    })

df_scraping_baru = pd.DataFrame(data_hotel_baru)

df_scraping_baru = df_scraping_baru.drop_duplicates(
    subset=["URL_Detail"]
).reset_index(drop=True)

print("\nJUMLAH DATA HOTEL:", len(df_scraping_baru))
display(df_scraping_baru)
```

```python

# STEP 34: MEMBERSIHKAN DATA HOTEL HASIL SCRAPING

# Menghapus data yang bukan hotel atau penginapan
kata_dikecualikan = [
    "wisata agro",
    "restaurant",
    "restoran",
    "kabin perkemahan"
]

df_scraping_baru["Nama_Hotel"] = (
    df_scraping_baru["Nama_Hotel"]
    .str.replace(r"^(Hotel Resor|Hotel)\s+[\d,]+\s*", "", regex=True)
    .str.strip()
)

df_scraping_baru = df_scraping_baru[
    ~df_scraping_baru["Nama_Hotel"].str.lower().str.contains(
        "|".join(kata_dikecualikan),
        na=False
    )
].copy()

df_scraping_baru = df_scraping_baru.reset_index(drop=True)

print("JUMLAH DATA SETELAH PEMBERSIHAN:", len(df_scraping_baru))
print("\nPEMERIKSAAN DATA KOSONG:")
print(df_scraping_baru.isnull().sum())

print("\nPEMERIKSAAN DUPLIKAT:")
print("Jumlah duplikat:", df_scraping_baru.duplicated().sum())

display(df_scraping_baru)
```

```python

# STEP 35: MENYIMPAN DATASET SCRAPING TIMES MALANG

import hashlib
import json
from datetime import datetime

# Menyimpan hasil scraping dalam CSV
file_csv = PROCESSED_DIR / "hotel_malang_timesmalang.csv"
df_scraping_baru.to_csv(file_csv, index=False, encoding="utf-8-sig")

# Menyimpan HTML mentah dari ketiga halaman
file_html = RAW_DIR / "timesmalang_hotel_raw.html"

with open(file_html, "w", encoding="utf-8") as f:
    for url in url_halaman:
        response = requests.get(
            url,
            headers=headers,
            timeout=30
        )
        f.write(f"\n<!-- SUMBER: {url} -->\n")
        f.write(response.text)
        f.write("\n")

# Menghitung ukuran dan MD5
def hitung_md5(file):
    return hashlib.md5(file.read_bytes()).hexdigest()

print("DATASET BERHASIL DISIMPAN")
print("CSV:", file_csv)
print("Jumlah data:", len(df_scraping_baru))
print("Ukuran CSV:", file_csv.stat().st_size, "byte")
print("MD5 CSV:", hitung_md5(file_csv))

print("\nRAW HTML")
print("File:", file_html)
print("Ukuran HTML:", file_html.stat().st_size, "byte")
print("MD5 HTML:", hitung_md5(file_html))
```
